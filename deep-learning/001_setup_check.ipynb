## The sanity check

Three things every notebook in this track relies on: a matrix product, a sigmoid that
does not overflow, and reading a CSV out of `../data/`.

```python
import numpy as np

W = np.array([[0.2, -0.5], [0.7, 0.1], [-0.3, 0.9]])   # (3, 2)
x = np.array([1.0, 2.0])                                # (2,)
print('W @ x =', W @ x, 'shape', (W @ x).shape)
assert (W @ x).shape == (3,)

def sigmoid(z):
    # The stable form. exp(1000) overflows; this does not.
    return np.where(z >= 0, 1 / (1 + np.exp(-np.abs(z))),
                    np.exp(-np.abs(z)) / (1 + np.exp(-np.abs(z))))

print('sigmoid([-1000, -1, 0, 1, 1000]) =',
      np.round(sigmoid(np.array([-1000., -1., 0., 1., 1000.])), 4))
assert np.isfinite(sigmoid(np.array([-1000., 1000.]))).all()
print('\nno overflow warnings above - good')
```

```python
import pandas as pd

df = pd.read_csv('../data/placement.csv')
print(df.shape)
print(df.head())
print('\nclass balance:\n', df['placed'].value_counts())
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(4.5, 3.6))
for label, marker in [(0, 'o'), (1, '^')]:
    sub = df[df['placed'] == label]
    ax.scatter(sub['cgpa'], sub['resume_score'], marker=marker,
               label=f'placed={label}', alpha=0.8)
ax.set(xlabel='cgpa', ylabel='resume_score', title='placement.csv')
ax.legend(); plt.tight_layout(); plt.show()
```

If you see a scatter plot with two visibly different groups, you are set up.
