## Part A — Is your environment ready?

The practical lessons need Python 3.9 or newer. Later lessons add `langchain` and friends;
this cell only checks what is needed to start.

```python
import sys, platform

print(f"Python  : {sys.version.split()[0]}")
print(f"Platform: {platform.system()} {platform.release()}")

major, minor = sys.version_info[:2]
ok = (major, minor) >= (3, 9)
print(f"\nPython 3.9 or newer: {'yes' if ok else 'NO - please upgrade'}")

for pkg in ["numpy", "sklearn"]:
    try:
        m = __import__(pkg)
        print(f"  {pkg:<10} {getattr(m, '__version__', 'installed')}")
    except ImportError:
        print(f"  {pkg:<10} missing  (needed from lesson 003)")
```

## Part B — The module, laid out

Three parts, roughly 17 lessons. Fundamentals come first because both RAG and agents are
built out of the same components.

```python
MODULE = {
    "Part 1 - Fundamentals": [
        "What LangChain is, and why it is needed",
        "Components overview",
        "Models",
        "Prompts",
        "Output parsing",
        "Runnables and LCEL",
        "Chains",
        "Memory",
    ],
    "Part 2 - RAG": [
        "Document loaders",
        "Text splitters",
        "Embeddings",
        "Vector databases",
        "Retrievers",
        "Build a RAG application from scratch",
    ],
    "Part 3 - Agents": [
        "Tools and toolkits",
        "Tool calling",
        "Build an AI agent",
    ],
}

n = 0
for part, topics in MODULE.items():
    print(f"\n{part}  ({len(topics)} lessons)")
    for t in topics:
        n += 1
        print(f"   {n:>2}. {t}")
print(f"\n{n} lessons in total")
```

## Part C — What this module touches

The argument for starting with LangChain is that it gives you a working view of the whole
user side at once, rather than one topic in depth.

```python
COVERAGE = [
    ("Closed-source models", "through LLM APIs"),
    ("Open-source models",   "through Hugging Face, or Ollama locally"),
    ("Prompt engineering",   "prompt templates and techniques"),
    ("RAG",                  "a complete application, built from scratch"),
    ("AI agents",            "tools, tool calling, and a working agent"),
    ("LLMOps",               "a first taste, not the full treatment"),
]
for area, how in COVERAGE:
    print(f"  {area:<22} {how}")

print("\nEach of these gets a module of its own later.")
print("This one is the first pass over all of them.")
```

## Part D — Pace

At about two lessons a week, roughly 17 lessons is about 8 weeks. The pace matters less than
leaving time to practise before the next idea arrives.

```python
lessons, per_week = 17, 2
weeks = lessons / per_week
print(f"{lessons} lessons at {per_week} a week -> {weeks:.1f} weeks (about {weeks/4.3:.1f} months)")

for rate in (1, 2, 3, 5):
    print(f"  {rate} a week -> {lessons/rate:>4.1f} weeks")
```

## What to take away

- **LangChain is an open-source framework** for building LLM applications out of modular
  components.
- **It is the first step on the user side**, and it touches everything on that side at once.
- **Three parts:** fundamentals, then RAG, then agents.
- **It targets version 0.3** and aims for the useful **80%** of the library.
- **The concepts matter more than the API** — runnables and chains are taught so that a future
  version is easy to pick up.
