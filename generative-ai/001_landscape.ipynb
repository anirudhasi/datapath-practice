# Generative AI 001 — The Landscape

This lesson has no model code. It builds a mental map, and the map is what the rest of the
course hangs on.

So this notebook is a drill, not a demo. It has one job: make the **user side / builder side**
split automatic, so that when you meet a new term later you already know where it goes.

Nothing here needs an API key or an internet connection.

## Part A — The eight terms from the lesson

Fill in your answer for each term before you scroll down. Write `"user"`, `"builder"` or
`"both"`.

```python
# Your answers. Edit these before running the next cell.
my_answers = {
    "prompt engineering": "?",
    "RAG":                "?",
    "AI agents":          "?",
    "vector databases":   "?",
    "pre-training":       "?",
    "RLHF":               "?",
    "quantization":       "?",
    "fine-tuning":        "?",
}
```

```python
KEY = {
    "prompt engineering": ("user",    "Refining the input to a model that already exists."),
    "RAG":                ("user",    "Question answering over your own private documents."),
    "AI agents":          ("user",    "An LLM plus tools, so it can act and not only talk."),
    "vector databases":   ("user",    "A component used inside RAG."),
    "pre-training":       ("builder", "Training a foundation model on huge data and hardware."),
    "RLHF":               ("builder", "Shaping how a model behaves while it is being created."),
    "quantization":       ("builder", "Optimising a model to run in different environments."),
    "fine-tuning":        ("both",    "A deep builder version, and a lighter user version."),
}

score = 0
for term, mine in my_answers.items():
    correct, why = KEY[term]
    ok = mine.strip().lower() == correct
    score += ok
    mark = "correct" if ok else f"-> {correct}"
    print(f"{term:<20} you said {mine:<9} {mark:<12} {why}")

print(f"\n{score} of {len(KEY)}")
```

The rule behind every answer: **are you working on the model, or working with it?**

- Working *with* a model that already exists → user side.
- Working *on* the model itself → builder side.

Fine-tuning is the one that lands in both, because there is a deep version done while building
a model and a lighter version done by people using one.

## Part B — Terms you have not met yet

These all appear later in the course. You have not been taught them. Guess anyway — the point
is to practise using the rule, not to already know the answers.

```python
LATER = {
    "LangChain":              ("user",    "A framework for building applications on an LLM."),
    "chunking":               ("user",    "Splitting your documents before putting them in a store."),
    "LangGraph":              ("user",    "Building agents as graphs, on top of existing models."),
    "knowledge distillation": ("builder", "Compressing a big model into a smaller one."),
    "tokenization strategy":  ("builder", "Chosen when a foundation model is pre-trained."),
    "PEFT":                   ("builder", "Parameter-efficient fine-tuning, done to the model."),
    "MCP":                    ("user",    "A protocol for giving a model access to tools and data."),
    "LLM-as-a-judge":         ("user",    "Evaluating the output of an application you built."),
}

for term, (correct, why) in LATER.items():
    print(f"{term:<24} {correct:<9} {why}")
```

Notice how much you can say about an unfamiliar term just from which box it falls into. If
something is user-side, you know you will need an existing model, an API or a local runtime,
and some ordinary software engineering. If it is builder-side, you know it needs training
data, hardware and a deep learning framework.

## Part C — Task-specific against general

The other idea worth making concrete: an ordinary ML model does one job, a foundation model
does many. This cell just writes that difference out.

```python
ordinary = {
    "name": "stock price predictor",
    "trained_on": "historical prices for one market",
    "can_do": ["predict a stock price"],
}
foundation = {
    "name": "a large language model",
    "trained_on": "roughly the whole internet",
    "can_do": ["generate text", "summarise", "classify sentiment",
               "answer questions", "write code"],
}

for m in (ordinary, foundation):
    print(f"{m['name']}")
    print(f"  trained on : {m['trained_on']}")
    print(f"  can do     : {len(m['can_do'])} task(s) -> {', '.join(m['can_do'])}")
    print()

print("Ask the stock model for a cricket score and you do not get a worse answer.")
print("You get no answer. It was never built to have one.")
```

## What to take away

- **Generative AI creates new content.** That is the break from older machine learning, which
  did regression, classification and ranking.
- **AI ⊃ machine learning ⊃ deep learning ⊃ generative AI.**
- **Foundation models sit at the centre of the map**: very large, trained on roughly the whole
  internet, and general rather than task-specific.
- **Every term is user-side or builder-side.** Ask: am I working *with* the model, or *on* it?
- **Fine-tuning is the exception** that belongs to both.
